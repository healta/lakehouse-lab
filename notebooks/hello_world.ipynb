## Spark + Delta + Unity Catalog

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
spark.version
```

```python
spark.sql("CREATE TABLE IF NOT EXISTS main.default.hello_jupyter (message STRING) USING DELTA")
spark.sql("INSERT INTO main.default.hello_jupyter VALUES ('Hello from Jupyter')")
spark.table("main.default.hello_jupyter").show()
```

```python
# A Python function running inside Spark
from pyspark.sql.functions import udf

spark.range(3).select(udf(lambda i: f"hello {i}")("id")).show()
```

## S3

```python
import boto3

s3 = boto3.client("s3")
s3.put_object(Bucket="lakehouse", Key="hello.txt", Body=b"Hello from S3")
s3.get_object(Bucket="lakehouse", Key="hello.txt")["Body"].read()
```

```python
# Spark reading the same file by its path
spark.read.text("s3://lakehouse/hello.txt").show()
```
